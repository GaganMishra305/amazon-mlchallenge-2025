```python
import pandas as pd
import re
import numpy as np

# NOTE: this cell applies the parser to the existing `new_d` dataframe in-place
# `new_d` should already be defined (for example, new_d = output)

# Load your dataset if new_d is not already defined
try:
    new_d
except NameError:
    new_d = pd.read_csv("dataset/train.csv")

def parse_catalog_content(text):
    """Parse a catalog_content text blob and return structured fields. Only assign a value if the label exists in the text, and extract the exact value/unit as they appear after the label. No fallback. All missing fields are None."""
    if pd.isna(text) or not str(text).strip():
        return {
            "item_name": None,
            "bullet_points": None,
            "product_description": None,
            "value": None,
            "unit": None
        }
    s = str(text)
    s = re.sub(r'\s+', ' ', s).strip()
    item_name = None
    bullet_points = None
    product_description = None
    value = None
    unit = None

    # Item Name: only if label exists, extract exact value
    m = re.search(r'Item Name[:\-\s]*([^\n\r]+?)(?=(?:Bullet Point|Bullets|Product Description|Description|Value|Unit|$))', s, re.I)
    if m:
        item_name = m.group(1).strip()

    # Bullet Points: only if label exists, extract all as one string
    bullet_matches = re.findall(r'Bullet Point\s*\d*[:\-]\s*([^\n\r]+)', s, re.I)
    if bullet_matches:
        bullet_points = " | ".join([b.strip() for b in bullet_matches if b.strip()])

    # Product Description: only if label exists, extract exact value
    m = re.search(r'Product Description[:\-\s]*([^\n\r]+?)(?=(?:Value|Unit|$))', s, re.I)
    if m:
        product_description = m.group(1).strip()

    # Value: only if label exists, extract exact value and unit (if present)
    m = re.search(r'Value[:\-\s]*([\d,\.]+)\s*([a-zA-Z%/\-]+)?', s, re.I)
    if m:
        val = m.group(1).replace(',', '')
        try:
            value = float(val)
        except Exception:
            value = m.group(1).strip()  # fallback: keep as string if not float
        if m.group(2):
            unit = m.group(2).strip()

    # Unit: only if label exists and not already set from Value
    if unit is None:
        m = re.search(r'Unit[:\-\s]*([^\n\r]+?)(?=(?:$|\s|\n|\r))', s, re.I)
        # if m:
        unit = m.group(1).strip()

    return {
        "item_name": item_name,
        "bullet_points": bullet_points,
        "product_description": product_description,
        "value": value,
        "unit": unit
    }
```

```python
# Apply parsing to the `new_d` dataframe in-place (do not create a new dataframe)
parsed_cols = new_d['catalog_content'].apply(parse_catalog_content).apply(pd.Series)
# Assign parsed columns back into new_d
for c in parsed_cols.columns:
    new_d[c] = parsed_cols[c]
# Save parsed dataset (optional)
new_d.to_csv("dataset/train_parsed_detailed.csv", index=False)
```

```python
# Remove outliers from the length columns and update new_d with new columns
def remove_outliers_iqr(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return series[(series >= lower) & (series <= upper)]

# Compute lengths
new_d['item_name_len'] = new_d['item_name'].astype(str).str.len()
new_d['bullet_points_len'] = new_d['bullet_points'].astype(str).str.len()
new_d['product_description_len'] = new_d['product_description'].astype(str).str.len()

# Remove outliers and set outliers to np.nan
for col in ['item_name_len', 'bullet_points_len', 'product_description_len']:
    lens = new_d[col]
    lens_no_out = remove_outliers_iqr(lens)
    # Set outliers to np.nan
    outlier_mask = ~lens.index.isin(lens_no_out.index)
    new_d.loc[outlier_mask, col] = np.nan

# Optionally, show summary
new_d[['item_name_len', 'bullet_points_len', 'product_description_len']].describe()
```

```python
# Ensure the length columns exist in new_d, recompute if missing or after filtering
if 'item_name_len' not in new_d.columns or 'bullet_points_len' not in new_d.columns or 'product_description_len' not in new_d.columns:
    new_d['item_name_len'] = new_d['item_name'].astype(str).str.len()
    new_d['bullet_points_len'] = new_d['bullet_points'].astype(str).str.len()
    new_d['product_description_len'] = new_d['product_description'].astype(str).str.len()
```

```python
# Create a DataFrame with only rows where all three length columns are within their respective IQRs
def within_iqr(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return (series >= lower) & (series <= upper)

mask_item = within_iqr(new_d['item_name_len'])
mask_bullet = within_iqr(new_d['bullet_points_len'])
mask_desc = within_iqr(new_d['product_description_len'])

mask_all_within = mask_item & mask_bullet & mask_desc
df_iqr = new_d[mask_all_within].reset_index(drop=True)

print(f"Rows within IQR for all three fields: {df_iqr.shape[0]}")
df_iqr.head()
```

```python
import re

def preprocess_text(text):
    """
    Preprocess text by lowering case, removing emojis, extra whitespace, and non-alphanumeric characters (except spaces).
    """
    if not isinstance(text, str):
        return ""
    # Remove emojis
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags (iOS)
        "\U00002700-\U000027BF"  # Dingbats
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    text = emoji_pattern.sub(r'', text)
    # Lowercase
    text = text.lower()
    # Remove non-alphanumeric characters except spaces
    text = re.sub(r'[^a-z0-9 ]+', ' ', text)
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
data['item_data'] = data['item_data'].apply(preprocess_text)
```
